# Guiding in prompts

```python
import os
from dotenv import load_dotenv, find_dotenv
from google import genai
from google.genai import types

load_dotenv(find_dotenv())

# Reads GEMINI_API_KEY from your .env file (or you can pass api_key=... explicitly)
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="What is LangChain? Explain in 2 bullet points.",
    config=types.GenerateContentConfig(
        system_instruction="You are a gen-z assistant who always answers in a fun way",
        temperature=0.7,
    ),
)

print(response.text)
```

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

# 1. Load your Gemini API key from .env
load_dotenv(find_dotenv())

# 2. Define llm_genai
llm_genai = ChatGoogleGenerativeAI(model="gemini-3.5-flash", temperature=0.7)
```

```python
from langchain_core.prompts import PromptTemplate
result=llm_genai.invoke("Tell me a joke.Give me the result in key-value pair format with the following step and punchline")
result.content
```

# Using Pydantic Models

```python
from pydantic import BaseModel,Field
class llm_schema(BaseModel):
    setup:str=Field(description="The setup for the joke")
    punchline:str=Field(description="The punchline for the joke")
```

```python
llm_structured_output=llm_genai.with_structured_output(llm_schema)
result=llm_structured_output.invoke("Tell me a joke")
result.punchline
```

# Using Typed Dict

```python
from typing import TypedDict
class llm_schema_td(TypedDict):
    setup:str
    punchline:str
```

```python
obj=llm_schema_td({"setup":"some setup","punchline":"some punchline"})
obj['setup']
```

```python
llm_structured_typed_output=llm_genai.with_structured_output(llm_schema_td)
result=llm_structured_typed_output.invoke("Tell me a joke")
result
```
