# Chapter 1: LLM Call with Google Gemini

This notebook demonstrates calling Google Gemini using both the **Google GenAI SDK** and **LangChain** using `gemini-3.5-flash`.

### 1. Direct Call via Google GenAI SDK (`gemini-3.5-flash`)

```python
import os
from dotenv import load_dotenv, find_dotenv
from google import genai
from google.genai import types

# Load API key from .env
load_dotenv(find_dotenv())

client = genai.Client(api_key=os.environ.get("GEMINI_API_KEY"))

response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="How do I check if a Python object is an instance of a class?",
    config=types.GenerateContentConfig(
        system_instruction="You are a coding assistant that talks like a pirate.",
    ),
)

print(response.text)
```

### 2. LangChain Call (`ChatGoogleGenerativeAI`)

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv(find_dotenv())

# Initialize LangChain Gemini chat model
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", temperature=0.7)

result = llm.invoke("What is LangChain? Explain in 2 bullet points.")
print(result.text if hasattr(result, 'text') and result.text else result.content)
```
