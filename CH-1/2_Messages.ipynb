# Chapter 1: LangChain Messages (`SystemMessage`, `HumanMessage`, `AIMessage`)

In LangChain, chat models take a sequence of **messages** instead of raw strings. This allows you to provide context, persona instructions, and conversation history.

### 1. Initialize the LLM (`llm_genai`)
> **Note:** You must define `llm_genai` in this notebook before calling it.

```python
import os
from dotenv import load_dotenv, find_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

# 1. Load API keys from .env
load_dotenv(find_dotenv())

# 2. Initialize the model (defining llm_genai)
llm_genai = ChatGoogleGenerativeAI(model="gemini-3.5-flash", temperature=0.7)
print("llm_genai initialized successfully!")
```

### 2. Basic Message Call with System & Human Messages

```python
messages = [
    SystemMessage(content="You are an expert Python and AI tutor."),
    HumanMessage(content="Explain the difference between SystemMessage and HumanMessage in LangChain in 2 bullet points."),
]

# Call llm_genai
response = llm_genai.invoke(messages)
print(response.text if hasattr(response, 'text') and response.text else response.content)
```

### 3. Multi-turn Conversation History (Contextual Memory)

```python
chat_history = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Hi, my name is Kalai."),
    AIMessage(content="Hello Kalai! Nice to meet you. How can I help you today?"),
    HumanMessage(content="What is my name?"),
]

response = llm_genai.invoke(chat_history)
print(response.text if hasattr(response, 'text') and response.text else response.content)
```
