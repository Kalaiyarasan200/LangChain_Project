```python
import os
from dotenv import load_dotenv, find_dotenv
from google import genai
from google.genai import types

load_dotenv(find_dotenv())

# Reads GEMINI_API_KEY from your .env file (or you can pass api_key=... explicitly)
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))
if os.environ.get("GEMINI_API_KEY"):
    print("Both API Key variable is exists")
else:
    raise ValueError("GEMINI_API_KEY not found")
from langchain_core.messages import HumanMessage,AIMessage,SystemMessage
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm_genai=ChatGoogleGenerativeAI(model="gemini-3.5-flash",temperature=0.7)
```

```python
#First Chain
#Task -1 [Prompt]
prompt_template=ChatPromptTemplate.from_messages([
    ("system","You are a helpful assistant"),
    ("human","{input}")
])
```

```python
#Task -2 [llm]
llm_genai=ChatGoogleGenerativeAI(model="gemini-3.5-flash",temperature=0.7)
```

```python
#Task -3[Strout parser]
str_parser=StrOutputParser()
```

```python
#Task -4 [Custom Runnable]
from langchain_core. runnables import RunnableLambda
def dictionary_maker(text:str)->dict:
    return {"text":text}
dictionary_maker_runnable=RunnableLambda(dictionary_maker)
```

# Parallel Chain 1

```python
#Task -1[prompt]
linkedin_prompt=ChatPromptTemplate.from_messages([
    ("system","You are a post generator"),
    ("human","Create a post for the following text for linkedin:{text}")
])

#Task -2 [llm]

llm_genai=ChatGoogleGenerativeAI(model="gemini-3.5-flash",temperature=0.7)

#Task -3[Strout parser]

str_parser=StrOutputParser()

chain_linkedin=linkedin_prompt | llm_genai | str_parser
```

# Parallel Chain 2

```python
from langchain_core.runnables import RunnableSequence,RunnableLambda,RunnableParallel
```

```python
def insta_chain(text:dict):
    text=text["text"]
    
    #Task -1[prompt]
    insta_prompt=ChatPromptTemplate.from_messages([
        ("system","You are a post generator"),
        ("human","Create a post for the following text for instagram:{text}")
    ])

    #Task -2 [llm]

    llm_genai=ChatGoogleGenerativeAI(model="gemini-3.5-flash",temperature=0.7)

    #Task -3[Strout parser]

    str_parser=StrOutputParser()

    chain_insta=insta_prompt | llm_genai | str_parser
    result=chain_insta.invoke(text)
    return result
insta_chain_runnable=RunnableLambda(insta_chain)
```

# Final Orchestration

```python
from langchain_core.runnables import branch
final_chain=(
    prompt_template |
    llm_genai |
    str_parser |
    dictionary_maker_runnable |
    RunnableParallel(branches={"Linkedin":chain_linkedin,"instagram":insta_chain_runnable})
)
```

```python
final_chain.invoke("KGF")
```

# Chain as a Runnable

```python
# Task -1 Beautify Function
def beautify(final_response:dict)->dict:
    linkedin_response=final_response['branches']['Linkedin']
    instagram_response=final_response['branches']['instagram']

    return {"linkedin":linkedin_response,"instagram":instagram_response}
beautify_runnable=RunnableLambda(beautify)

# Task -2 [Final chain]

#final chain

# Beautified chain

beautified_chain=final_chain | beautify_runnable

beautified_chain.invoke("Pushpa")
```
