# Chains in Langchain

```python
import os
from dotenv import load_dotenv, find_dotenv
from google import genai
from google.genai import types

load_dotenv(find_dotenv())

# Reads GEMINI_API_KEY from your .env file (or you can pass api_key=... explicitly)
client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))
if os.environ.get("GEMINI_API_KEY"):
    print("Both API Key variable is exists")
else:
    raise ValueError("GEMINI_API_KEY not found")
from langchain_core.messages import HumanMessage,AIMessage,SystemMessage
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
llm_genai=ChatGoogleGenerativeAI(model="gemini-3.5-flash",temperature=0.7)
```

# First Chain

```python
#First Chain
#Task -1 [Prompt]
prompt_template=ChatPromptTemplate.from_messages([
    ("system","You are a helpful assistant"),
    ("human","{input}")
])
```

```python
#Task -2 [llm]
llm_genai=ChatGoogleGenerativeAI(model="gemini-3.5-flash",temperature=0.7)
```

```python
#Task -3[Strout parser]
str_parser=StrOutputParser()
```

# Manual Invocation

```python
template=prompt_template.invoke({"input":"What is the capital of france?"})
res=llm_genai.invoke(template)
final_result=res.content
```

# Chain Invocation

```python
chain =prompt_template | llm_genai |str_parser
chain.invoke({"input":"What is the capital of france?"})
```

```python
from langchain_core.runnables import RunnableSequence
chain_1=RunnableSequence(prompt_template,llm_genai,str_parser)
chain_1.invoke("what is the capital of Italy")
```
